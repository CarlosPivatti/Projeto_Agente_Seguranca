````python
from typing import TypedDict
````

````python
class AgentState(TypedDict):
  user_query: str
  web_answer: str
  final_answer: str
  router_decision: str
````

````python
def funcao_agente_web(state: AgentState) -> dict:
    """Função que executa a busca na web e extrai o texto puro da resposta."""
    system_prompt = "Você é um engenheiro de Segurança do Trabalho especialista. Use a ferramenta de busca para fornecer uma resposta técnica detalhada e estruturada em Markdown."
    agente = create_react_agent(model=llm, tools=tools, prompt=system_prompt)
    resultado = agente.invoke({"messages": [("user", state["user_query"])]})
    conteudo = resultado["messages"][-1].content

    # Extração forçada de texto para evitar que listas de dicionários quebrem a UI
    if isinstance(conteudo, list):
        texto_final = "".join([block.get('text', '') for block in conteudo if isinstance(block, dict) and 'text' in block])
    else:
        texto_final = str(conteudo)

    return {"web_answer": texto_final.strip()}
````

````python
# Testando a função do agente isoladamente
meu_estado = {"user_query": "Quais os riscos da indústria metalúrgica?", "web_answer": "", "final_answer": ""}

# Chamando a função e exibindo apenas o resultado da busca web
resultado_do_estado = funcao_agente_web(meu_estado)
print("Resposta da Web coletada:")
print(resultado_do_estado["web_answer"])
````

````python
from langgraph.graph import START, StateGraph, END
````

````python
workflow = StateGraph(AgentState)
````

````python
# Reiniciamos o workflow para evitar o erro de nó duplicado ao re-executar a célula
workflow = StateGraph(AgentState)

workflow.add_node("agente_web", funcao_agente_web)
workflow.add_edge(START, "agente_web")
workflow.add_edge("agente_web", END)

app = workflow.compile()
````

````python
def router_agent(state: AgentState) -> dict:
    """ Agente roteador que decide qual ferramenta usar. """
    router_prompt = """
    Você é um classificador de intenções para um sistema de Segurança do Trabalho.
    Sua única tarefa é responder 'web_search' para perguntas técnicas ou 'outro' para saudações.

    Pergunta: {user_query}

    Regras:
    - Se a pergunta envolver qualquer risco, norma (NR), ergonomia, metalurgia ou termo industrial, responda APENAS: web_search
    - Se for apenas 'oi', 'olá', 'quem é você' ou conversa genérica, responda APENAS: outro
    """

    router_agent_instance = create_react_agent(model=llm, tools=[], prompt=router_prompt)
    response = router_agent_instance.invoke({"messages": [("user", state["user_query"])]})
    content = response['messages'][-1].content

    if isinstance(content, list):
        text = " ".join([block.get('text', '') for block in content if isinstance(block, dict)])
    else:
        text = str(content)

    decision = text.strip().lower()

    # Palavras-chave de salvaguarda (Backstop logic)
    keywords = ["metalurgia", "risco", "ergonomia", "quimica", "segurança", "trabalho", "nr", "industrial", "doença", "acidente", "fumos"]
    query_lower = state["user_query"].lower()

    if "web_search" in decision or any(w in query_lower for w in keywords):
        return {"router_decision": "web_search"}
    return {"router_decision": "outro"}
````

````python
workflow = StateGraph(AgentState)

workflow.add_node("agente_web", funcao_agente_web)
workflow.add_node("router", router_agent)

workflow.add_edge(START, "router")

workflow.add_conditional_edges(
    "router",
    lambda state: state["router_decision"], # Lógica baseada no valor armazenado no estado
    {
        "web_search": "agente_web",
        "outro": END
    }
)

workflow.add_edge("agente_web", END)

app = workflow.compile()
````
